# Quantum Emitters and LDOS Enhancement

Computes LDOS at edge and bulk sites via the retarded Green's function, verifies
non-negativity, and quantifies the enhancement factor.


```python
import sys
sys.path.insert(0, "..")   # find src/nh_topo from notebooks/

import numpy as np
import matplotlib.pyplot as plt
import nh_topo as nt
from nh_topo.config import load_all, get_physics_params, get_figure_dir
from pathlib import Path

cfg = load_all(config_dir="../config")
p   = get_physics_params(cfg)

# Quick confirmation
print(f"t1={p['t1']}, t2={p['t2']}, gamma={p['gamma']}, N={p['N']}")
print(f"Topological: t2/t1 = {p['t2']/p['t1']:.2f} > 1 → |nu|={nt.topological_invariant(p['t1'], p['t2'])}")
```

```python
H      = nt.build_ssh(p['N'], p['t1'], p['t2'], p['gamma'])
omegas = np.linspace(p['om_min'], p['om_max'], p['n_omega'])
gap_oms = np.linspace(-p['gap_hw'], p['gap_hw'], 120)

edge_s = p['edge_site']
bulk_s = p['bulk_site']

rho_edge = nt.ldos_spectrum(H, edge_s, omegas,   p['eta'])
rho_bulk = nt.ldos_spectrum(H, bulk_s, omegas,   p['eta'])
rho_e_gap = nt.ldos_spectrum(H, edge_s, gap_oms, p['eta'])
rho_b_gap = nt.ldos_spectrum(H, bulk_s, gap_oms, p['eta'])

assert np.all(rho_edge >= -1e-10), "LDOS must be non-negative"
assert np.all(rho_bulk >= -1e-10), "LDOS must be non-negative"

E_int = nt.enhancement_factor(H, edge_s, bulk_s, gap_oms, p['eta'])
print(f"Integrated LDOS enhancement: {E_int:.2f}x")
print(f"LDOS(edge, omega=0) = {nt.ldos_at(H, edge_s, 0.0, p['eta']):.3f}")
print(f"LDOS(bulk, omega=0) = {nt.ldos_at(H, bulk_s, 0.0, p['eta']):.4f}")

rho_spatial = nt.ldos_spatial_map(H, 0.0, p['eta'])
```

```python
fig, axes = nt.plotting.plot_ldos_edge_vs_bulk(
    omegas, rho_edge, rho_bulk, gap_oms,
    rho_e_gap, rho_b_gap, E_int, edge_s, bulk_s)
nt.plotting.save_fig(fig, Path('..') / get_figure_dir(cfg, '05_emitters_ldos_nanophotonics'), "01_ldos_spectra")
plt.show()

fig, ax = nt.plotting.plot_ldos_spatial(rho_spatial, p['N'], edge_s, bulk_s)
nt.plotting.save_fig(fig, Path('..') / get_figure_dir(cfg, '05_emitters_ldos_nanophotonics'), "02_ldos_spatial")
plt.show()
```
