# 07 — Parameter-space maps

The enhancement onsets at the phase boundary `t2/t1=1`, is **largest in the
Hermitian limit `gamma=0`**, and is monotonically suppressed by loss. The
`eta`-independent edge weight `W0` is a clean topological order parameter
(`gamma`-independent; zero in the trivial phase).

```python
import sys; sys.path.insert(0, "../src")
import numpy as np, matplotlib.pyplot as plt
import nh_topo as nt
from nh_topo import plotting as P
cfg = nt.load_all(config_dir="../config"); p = nt.get_physics_params(cfg)
from pathlib import Path
def figdir(key): return Path("..") / nt.get_figure_dir(cfg, key)
print(f"t1={p['t1']}, t2={p['t2']}, gamma={p['gamma']}, N={p['N']}, "
      f"|nu|={nt.topological_invariant(p['t1'], p['t2'])}")
```

```python
ps = cfg["simulation"]["parameter_sweeps"]
r = np.linspace(ps["r_vals"]["min"], ps["r_vals"]["max"], ps["r_vals"]["num"])
g = np.linspace(ps["gamma_vals"]["min"], ps["gamma_vals"]["max"], ps["gamma_vals"]["num"])
gap = np.linspace(-p["gap_hw"], p["gap_hw"], 61)
enh = nt.enhancement_map(p["N"], r, g, gap, p["eta"]); w0 = nt.edge_weight_map(p["N"], r, g)
slices = [(0.0, P.C["topo"]), (0.3, P.C["edge"]), (0.6, P.C["bulk"])]
fig, _ = P.plot_sweeps(r, g, enh, w0, slices)
P.save_fig(fig, figdir("07_parameter_sweep_topological_ldos"), "01_maps"); plt.show()
```
