# Parameter-Space Map of Topological LDOS Enhancement

Sweeps $(t_2/t_1,\,\gamma/t_1)$ on a 30×30 grid and maps the integrated
LDOS enhancement factor, revealing the topological phase boundary.


```python
import sys
sys.path.insert(0, "..")   # find src/nh_topo from notebooks/

import numpy as np
import matplotlib.pyplot as plt
import nh_topo as nt
from nh_topo.config import load_all, get_physics_params, get_figure_dir
from pathlib import Path

cfg = load_all(config_dir="../config")
p   = get_physics_params(cfg)

# Quick confirmation
print(f"t1={p['t1']}, t2={p['t2']}, gamma={p['gamma']}, N={p['N']}")
print(f"Topological: t2/t1 = {p['t2']/p['t1']:.2f} > 1 → |nu|={nt.topological_invariant(p['t1'], p['t2'])}")
```

```python
ps      = cfg['simulation']['parameter_sweeps']
r_vals  = np.linspace(ps['r_vals']['min'],     ps['r_vals']['max'],     ps['r_vals']['num'])
g_vals  = np.linspace(ps['gamma_vals']['min'], ps['gamma_vals']['max'], ps['gamma_vals']['num'])
gap_oms = np.linspace(-p['gap_hw'], p['gap_hw'], 50)

print(f"Grid: {len(r_vals)} x {len(g_vals)} = {len(r_vals)*len(g_vals)} points")
print("Running sweep (may take ~2 min)...")

enh = nt.enhancement_map(p['N'], r_vals, g_vals, gap_oms, p['eta'])
print(f"Enhancement range: [{enh.min():.2f}, {np.percentile(enh,97):.2f}] (97th pct)")
```

```python
g_slices = [(0.0, "#2ca02c"), (0.2, "#1f77b4"), (0.4, "#ff7f0e"), (0.6, "#d62728")]
r_slices = [(1.2, "#1f77b4"), (1.5, "#2ca02c"), (2.0, "#d62728")]

fig, axes = nt.plotting.plot_enhancement_map(r_vals, g_vals, enh, g_slices, r_slices)
nt.plotting.save_fig(fig, Path('..') / get_figure_dir(cfg, '07_parameter_sweep_topological_ldos'), "01_enhancement_map")
plt.show()
```
